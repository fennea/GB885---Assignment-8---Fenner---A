```python
#input parameters
sales_per_unit = 40
cost_per_unit = 25

seasonality_factors_q1 = 0.9
seasonality_factors_q2 = 1.1
seasonality_factors_q3 = 0.8
seasonality_factors_q4 = 1.2

salesforce_expenses_q1 = 8000
salesforce_expenses_q2 = 8000
salesforce_expenses_q3 = 9000
salesforce_expenses_q4 = 9000

overhead_rate = 0.15

k = 35
b = 3000
```

```python
#advertising spend
advert_spend_q1 = 10000
advert_spend_q2 = 10000
advert_spend_q3 = 10000
advert_spend_q4 = 10000
```

```python
#gross profit calculations
sales_quantity_q1 = k * seasonality_factors_q1 * (b + advert_spend_q1) ** 0.5
sales_quantity_q2 = k * seasonality_factors_q2 * (b + advert_spend_q2) ** 0.5
sales_quantity_q3 = k * seasonality_factors_q3 * (b + advert_spend_q3) ** 0.5
sales_quantity_q4 = k * seasonality_factors_q4 * (b + advert_spend_q4) ** 0.5

COGS_q1 = cost_per_unit * sales_quantity_q1
COGS_q2 = cost_per_unit * sales_quantity_q2
COGS_q3 = cost_per_unit * sales_quantity_q3
COGS_q4 = cost_per_unit * sales_quantity_q4

sales_rev_q1 = sales_per_unit * sales_quantity_q1
sales_rev_q2 = sales_per_unit * sales_quantity_q2
sales_rev_q3 = sales_per_unit * sales_quantity_q3
sales_rev_q4 = sales_per_unit * sales_quantity_q4

gross_profit_q1 = sales_rev_q1 - COGS_q1
gross_profit_q2 = sales_rev_q2 - COGS_q2
gross_profit_q3 = sales_rev_q3 - COGS_q3
gross_profit_q4 = sales_rev_q4 - COGS_q4

annual_gross_profit = gross_profit_q1 + gross_profit_q2 + gross_profit_q3 + gross_profit_q4
```

```python
annual_gross_profit
```

```python
#operating expense calculations
overhead_q1 = overhead_rate * sales_rev_q1
overhead_q2 = overhead_rate * sales_rev_q2
overhead_q3 = overhead_rate * sales_rev_q3
overhead_q4 = overhead_rate * sales_rev_q4

operating_expenses_q1 = overhead_q1 + salesforce_expenses_q1 + advert_spend_q1
operating_expenses_q2 = overhead_q2 + salesforce_expenses_q2 + advert_spend_q2
operating_expenses_q3 = overhead_q3 + salesforce_expenses_q3 + advert_spend_q3
operating_expenses_q4 = overhead_q4 + salesforce_expenses_q4 + advert_spend_q4

annual_operating_expense = operating_expenses_q1 + operating_expenses_q2 + operating_expenses_q3 + operating_expenses_q4
annual_operating_expense
```

```python
#annual operating profit
annual_operating_profit = annual_gross_profit - annual_operating_expense
annual_operating_profit
```
